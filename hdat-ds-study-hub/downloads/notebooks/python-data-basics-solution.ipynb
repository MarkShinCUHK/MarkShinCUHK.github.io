## 3. if, for, function

if는 조건에 따라 코드를 선택합니다. for는 여러 값을 하나씩 반복합니다.

function은 자주 쓰는 계산에 이름을 붙인 것입니다.

- def로 정의합니다.
- 매개변수는 함수가 받을 입력입니다.
- return은 함수 밖으로 돌려줄 결과입니다.
- 들여쓰기는 Python 문법의 일부이므로 공백 수를 맞춰야 합니다.

```python
temperatures = [18.0, 24.5, 31.0]
labels = []

for temperature in temperatures:
    if temperature >= 30:
        labels.append("hot")
    else:
        labels.append("normal")

print(labels)
assert labels == ["normal", "normal", "hot"]
```

```python
def km_per_kwh(distance_km, energy_kwh):
    """거리와 사용 전력량으로 전비를 계산합니다."""
    if energy_kwh <= 0:
        raise ValueError("energy_kwh는 0보다 커야 합니다.")
    return distance_km / energy_kwh

result = km_per_kwh(180, 30)
print("함수 결과:", result)

assert result == 6.0
```

## 4. NumPy: 숫자 배열 다루기

NumPy는 같은 종류의 숫자를 빠르게 계산하는 라이브러리입니다. 보통 np라는 짧은 이름으로 불러옵니다.

핵심 확인 항목

- shape: 각 축의 길이
- ndim: 축의 개수
- size: 전체 원소 수
- dtype: 원소 자료형

머신러닝에서는 흔히 shape가 (행 개수, 특성 개수)인 2차원 배열을 사용합니다.

```python
import numpy as np

vehicle_array = np.array([
    [1200.0, 14.5, 2022],
    [1450.0, 12.8, 2023],
    [1680.0, 11.9, 2024],
    [1320.0, 13.7, 2025],
])

print("배열:\n", vehicle_array)
print("shape:", vehicle_array.shape)
print("ndim:", vehicle_array.ndim)
print("size:", vehicle_array.size)
print("dtype:", vehicle_array.dtype)
```

```python
first_row = vehicle_array[0]
weight_column = vehicle_array[:, 0]
efficiency_block = vehicle_array[1:3, 1]

print("첫 행:", first_row)
print("중량 열:", weight_column)
print("두 차량 전비:", efficiency_block)

assert first_row.shape == (3,)
assert weight_column.tolist() == [1200.0, 1450.0, 1680.0, 1320.0]
assert efficiency_block.tolist() == [12.8, 11.9]
```

### axis를 눈으로 이해하기

2차원 배열에서

- axis=0으로 줄이면 행 방향으로 내려가며 계산하여 열마다 결과가 하나씩 남습니다.
- axis=1로 줄이면 열 방향으로 옆으로 계산하여 행마다 결과가 하나씩 남습니다.

외우기 어렵다면 계산 뒤 남는 shape를 항상 출력하세요.

```python
column_means = vehicle_array.mean(axis=0)
row_means = vehicle_array.mean(axis=1)

print("열별 평균:", column_means)
print("열별 평균 shape:", column_means.shape)
print("행별 평균:", row_means)
print("행별 평균 shape:", row_means.shape)

assert column_means.shape == (3,)
assert row_means.shape == (4,)
assert np.isclose(column_means[0], 1412.5)
```

### reshape와 -1

reshape는 값을 바꾸지 않고 모양만 바꿉니다. 전체 원소 수는 전후가 같아야 합니다.

- reshape(2, 6): 2행 6열
- reshape(-1, 1): 행 개수는 NumPy가 계산하고 1열로 만듭니다.
- flatten 또는 reshape(-1): 1차원으로 펼칩니다.

```python
numbers = np.arange(12)
matrix_3x4 = numbers.reshape(3, 4)
column_vector = numbers.reshape(-1, 1)
flat_again = matrix_3x4.reshape(-1)

print("3 x 4:\n", matrix_3x4)
print("열 벡터 shape:", column_vector.shape)
print("다시 펼친 shape:", flat_again.shape)

assert matrix_3x4.shape == (3, 4)
assert column_vector.shape == (12, 1)
assert np.array_equal(numbers, flat_again)
```

### 브로드캐스팅

배열과 스칼라, 또는 호환되는 shape의 배열을 계산하면 NumPy가 작은 쪽을 자동으로 늘려 계산합니다.

열별 표준화 공식은 (값 - 열 평균) / 열 표준편차입니다. 평균과 표준편차 shape가 (열 개수,)이면 각 행에 자동 적용됩니다.

```python
features = vehicle_array[:, :2]
feature_mean = features.mean(axis=0)
feature_std = features.std(axis=0)
standardized = (features - feature_mean) / feature_std

print("평균:", feature_mean)
print("표준편차:", feature_std)
print("표준화 결과:\n", np.round(standardized, 3))
print("표준화 후 열 평균:", np.round(standardized.mean(axis=0), 7))

assert standardized.shape == features.shape
assert np.allclose(standardized.mean(axis=0), 0.0)
assert np.allclose(standardized.std(axis=0), 1.0)
```

### copy가 필요한 이유

NumPy 슬라이스는 원본과 메모리를 공유하는 view일 수 있습니다. 슬라이스를 수정했는데 원본까지 바뀌는 실수를 피하려면 독립 수정용 배열에 copy()를 사용합니다.

```python
original = np.array([10, 20, 30, 40])
view_part = original[:2]
view_part[0] = 999
print("view 수정 뒤 원본:", original)

safe_original = np.array([10, 20, 30, 40])
copied_part = safe_original[:2].copy()
copied_part[0] = 999
print("copy 수정 뒤 원본:", safe_original)

assert original[0] == 999
assert safe_original[0] == 10
```

## 5. pandas: 표 데이터 다루기

pandas의 DataFrame은 행과 열로 된 표입니다. 보통 pd라는 이름으로 불러옵니다.

이제 작은 가상 차량 데이터를 직접 만듭니다. np.nan은 빠진 값, 즉 결측치를 뜻합니다.

```python
import pandas as pd

vehicles = pd.DataFrame({
    "vehicle_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "model": ["IONIQ5", "KONA", "EV6", "KONA", "NEXO", "IONIQ5", "EV6", "NEXO"],
    "region": ["Seoul", "Busan", "Seoul", "Daejeon", "Busan", "Seoul", "Daejeon", "Busan"],
    "year": [2022, 2023, 2024, 2022, 2025, 2024, 2023, 2025],
    "mileage_km": [42000, 21000, 12000, 51000, 7000, 18000, 33000, 9000],
    "efficiency": [5.1, 5.7, 4.9, np.nan, 6.3, 5.4, np.nan, 6.1],
    "is_electric": [True, True, True, True, False, True, True, False],
})

print(vehicles)
print("shape:", vehicles.shape)
```

### head, info, dtypes, describe

데이터를 받으면 모델부터 만들지 말고 먼저 구조를 확인합니다.

info()는 내용을 직접 출력하므로 print로 다시 감싸지 않습니다.

```python
print("앞 3행")
print(vehicles.head(3))

print("\n정보")
vehicles.info()

print("\n자료형")
print(vehicles.dtypes)

print("\n수치 요약")
print(vehicles.describe())
```

### 열과 행 선택

- df["열"]: Series 한 개
- df[["열1", "열2"]]: DataFrame
- loc[행 조건, 열 이름]: 이름 기반
- iloc[행 위치, 열 위치]: 정수 위치 기반

Series는 1차원, DataFrame은 2차원입니다. 대괄호 개수에 따라 shape가 달라질 수 있습니다.

```python
models_series = vehicles["model"]
two_columns = vehicles[["model", "mileage_km"]]
named_selection = vehicles.loc[0:2, ["model", "efficiency"]]
position_selection = vehicles.iloc[:3, [1, 5]]

print("Series shape:", models_series.shape)
print("DataFrame shape:", two_columns.shape)
print("loc 결과:\n", named_selection)
print("iloc 결과:\n", position_selection)

assert models_series.ndim == 1
assert two_columns.shape == (8, 2)
assert named_selection.equals(position_selection)
```

### boolean mask로 조건 필터링

각 행이 조건을 만족하는지 계산하면 True/False Series가 됩니다. 이 mask를 DataFrame에 넣어 원하는 행만 고릅니다.

여러 조건을 결합할 때 pandas에서는 and/or 대신 괄호와 &, |를 사용합니다.

```python
mask = (vehicles["year"] >= 2024) & (vehicles["mileage_km"] < 20000)
recent_low_mileage = vehicles.loc[
    mask,
    ["vehicle_id", "model", "year", "mileage_km"],
]

print("mask:\n", mask)
print("조건에 맞는 차량:\n", recent_low_mileage)

assert mask.dtype == bool
assert recent_low_mileage["vehicle_id"].tolist() == [103, 105, 106, 108]
```

### 결측치 확인과 처리

- isna(): 값이 빠졌는지 확인
- isna().sum(): 열별 결측치 개수
- fillna(값): 결측치 채우기
- dropna(): 결측 행 또는 열 제거

여기서는 원본 DataFrame을 보존하기 위해 copy()한 뒤, efficiency의 결측치를 전체 중앙값으로 채웁니다. 실제 시험에서는 훈련 데이터로 계산한 값을 검증/테스트에 적용해야 데이터 누수를 피할 수 있습니다.

```python
missing_counts = vehicles.isna().sum()
median_efficiency = vehicles["efficiency"].median()

vehicles_clean = vehicles.copy()
vehicles_clean["efficiency"] = vehicles_clean["efficiency"].fillna(median_efficiency)

print("결측치 수:\n", missing_counts)
print("중앙값:", median_efficiency)
print("처리 후 결측치 수:", vehicles_clean["efficiency"].isna().sum())

assert missing_counts["efficiency"] == 2
assert np.isclose(median_efficiency, 5.55)
assert vehicles_clean["efficiency"].isna().sum() == 0
assert vehicles["efficiency"].isna().sum() == 2
```

### 파생 열, 정렬, groupby

파생 열은 기존 열로 계산한 새 열입니다. groupby는 같은 범주의 행을 묶어 집계합니다.

흐름은 split → apply → combine으로 이해하면 됩니다.

1. region별로 나눕니다.
2. mileage_km에는 평균, vehicle_id에는 개수를 적용합니다.
3. 결과를 하나의 표로 합칩니다.

```python
vehicles_clean["mileage_10k"] = vehicles_clean["mileage_km"] / 10_000

region_summary = (
    vehicles_clean
    .groupby("region", as_index=False)
    .agg(
        mean_mileage=("mileage_km", "mean"),
        vehicle_count=("vehicle_id", "count"),
        mean_efficiency=("efficiency", "mean"),
    )
    .sort_values("mean_mileage", ascending=False)
    .reset_index(drop=True)
)

print(region_summary)

assert set(region_summary["region"]) == {"Seoul", "Busan", "Daejeon"}
assert region_summary["vehicle_count"].sum() == len(vehicles_clean)
assert "mileage_10k" in vehicles_clean.columns
```

### merge로 두 표 결합

merge는 공통 키를 기준으로 두 표를 연결합니다. 데이터베이스의 join과 같은 개념입니다.

- how="left": 왼쪽 행은 모두 유지
- on="model": model 열을 키로 사용
- validate="many_to_one": 왼쪽에는 같은 모델이 여러 번, 오른쪽에는 모델당 한 행인지 검사

```python
model_lookup = pd.DataFrame({
    "model": ["IONIQ5", "KONA", "EV6", "NEXO"],
    "segment": ["SUV", "SUV", "Crossover", "SUV"],
    "powertrain": ["BEV", "BEV", "BEV", "FCEV"],
})

print(model_lookup)
```

```python
vehicles_enriched = vehicles_clean.merge(
    model_lookup,
    on="model",
    how="left",
    validate="many_to_one",
)

print(vehicles_enriched.head())
print("결합 후 shape:", vehicles_enriched.shape)

assert len(vehicles_enriched) == len(vehicles_clean)
assert vehicles_enriched["segment"].isna().sum() == 0
assert vehicles_enriched.loc[vehicles_enriched["model"] == "NEXO", "powertrain"].eq("FCEV").all()
```

### pandas에서 안전하게 수정하기

필터 결과를 수정할 때는 .copy()를 붙이는 습관이 안전합니다. 그렇지 않으면 SettingWithCopyWarning이 발생하거나 원본과 사본 중 무엇이 바뀌는지 헷갈릴 수 있습니다.

```python
seoul_vehicles = vehicles_enriched.loc[
    vehicles_enriched["region"] == "Seoul"
].copy()

seoul_vehicles["mileage_level"] = np.where(
    seoul_vehicles["mileage_km"] >= 30000,
    "high",
    "low",
)

print(seoul_vehicles[["vehicle_id", "mileage_km", "mileage_level"]])

assert seoul_vehicles["mileage_level"].tolist() == ["high", "low", "low"]
```

## 6. matplotlib: 숫자를 그림으로 확인하기

시각화는 데이터 분포, 이상치, 범주 차이를 빠르게 확인하는 도구입니다.

기본 순서

1. fig, ax = plt.subplots()로 도화지와 축을 만듭니다.
2. ax.bar, ax.scatter, ax.plot 등으로 그립니다.
3. 제목과 축 이름을 붙입니다.
4. plt.show()로 표시합니다.

시험 환경에 따라 한글 폰트가 없을 수 있으므로 그래프 제목은 영문을 사용합니다.

```python
import matplotlib.pyplot as plt

plot_data = (
    vehicles_clean
    .groupby("model", as_index=False)["mileage_km"]
    .mean()
    .sort_values("mileage_km")
)

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(plot_data["model"], plot_data["mileage_km"], color="#2563eb")
ax.set_title("Mean mileage by model")
ax.set_xlabel("Model")
ax.set_ylabel("Mileage (km)")
ax.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()

print(plot_data)
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(
    vehicles_clean["mileage_km"],
    vehicles_clean["efficiency"],
    c=vehicles_clean["year"],
    cmap="viridis",
    s=70,
)
ax.set_title("Mileage vs. efficiency")
ax.set_xlabel("Mileage (km)")
ax.set_ylabel("Efficiency")
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()

assert len(ax.collections[0].get_offsets()) == len(vehicles_clean)
```

## 7. pandas에서 PyTorch Tensor로

PyTorch 모델은 보통 float32 Tensor를 입력으로 받습니다.

변환 순서

1. 모델에 넣을 수치 열을 고릅니다.
2. pandas DataFrame을 to_numpy(dtype=np.float32)로 NumPy 배열로 바꿉니다.
3. torch.from_numpy 또는 torch.tensor로 Tensor를 만듭니다.
4. shape, dtype, device를 확인합니다.

문자열 범주형 열은 그대로 Tensor로 바꿀 수 없습니다. 이후 강의에서 one-hot encoding 또는 embedding을 배웁니다.

```python
import torch

feature_columns = ["year", "mileage_km", "efficiency"]
X_np = vehicles_clean[feature_columns].to_numpy(dtype=np.float32)
y_np = vehicles_clean["is_electric"].astype(np.float32).to_numpy()

X_tensor = torch.from_numpy(X_np)
y_tensor = torch.from_numpy(y_np).reshape(-1, 1)

print("X:", X_tensor.shape, X_tensor.dtype, X_tensor.device)
print("y:", y_tensor.shape, y_tensor.dtype, y_tensor.device)
print("첫 입력:", X_tensor[0])

assert X_tensor.shape == (8, 3)
assert y_tensor.shape == (8, 1)
assert X_tensor.dtype == torch.float32
assert y_tensor.dtype == torch.float32
assert X_tensor.device.type == "cpu"
assert torch.isfinite(X_tensor).all()
```

### TensorDataset과 DataLoader 맛보기

TensorDataset은 X와 y의 같은 인덱스를 한 묶음으로 관리합니다. DataLoader는 여러 행을 작은 배치로 나눠 모델에 전달합니다.

shuffle=False를 사용하면 처음에는 원래 순서를 그대로 확인할 수 있습니다. 실제 훈련에서는 보통 훈련 DataLoader에 shuffle=True를 사용합니다.

```python
from torch.utils.data import DataLoader, TensorDataset

dataset = TensorDataset(X_tensor, y_tensor)
loader = DataLoader(dataset, batch_size=3, shuffle=False)

batch_X, batch_y = next(iter(loader))
print("첫 배치 X shape:", batch_X.shape)
print("첫 배치 y shape:", batch_y.shape)
print("첫 배치 X:\n", batch_X)

assert len(dataset) == 8
assert batch_X.shape == (3, 3)
assert batch_y.shape == (3, 1)
assert batch_X.device.type == "cpu"
```
